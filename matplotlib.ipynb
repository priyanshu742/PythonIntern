```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

x=np.array([1,2,3,4])
y=np.array([1,4,9,16])
plt.plot(x,y,'r')#'r' is the color red
plt.xlabel('x')
plt.ylabel('y')
plt.title('title')
plt.show()

```

```python
plt.subplot(1,2,1)
plt.plot(x,y,'r--')
plt.subplot(1,2,2)
plt.plot(y,x,'g*-')
plt.show()
```

```python
labels='Batch A','Batch B','Batch C'
sizes=[10,30,45]
plt.figure(figsize=(6,6))
plt.pie(sizes,labels=labels,autopct='%1.0f%%',shadow=True,startangle=180)
plt.title('Pie Chart - AIML Batch Size')
plt.axis('equal')
plt.show()
```

```python
data={
    'day':['thur','fri','sat','sun','thur','fri','sat','sun'],
    'total_bill':[20.5,30.1,25.3,22.7,19.3,28.4,24.9,27.2],
    'tip':[3.5,4.1,3.3,3.7,3.1,4.0,3.4,3.8],
    'sex':['male','female','male','female','male','female','male','female'],
    'smoker':['yes','no','yes','no','yes','no','yes','no'],
}
df=pd.DataFrame(data)
print(df)
```

```python
category_counts=df['day'].value_counts()
print(category_counts)
```

```python
plt.figure(figsize=(3,2))
plt.bar(category_counts.index,category_counts.values,color=['yellow','blue'])
plt.xlabel('sex')
plt.ylabel('Count')
plt.title('Categorical Data-Count of sex')
plt.show()
```

```python
x=df['total_bill']
y=df['tip']
m,b=np.polyfit(x,y,1)
plt.figure(figsize=(6,4))
plt.scatter(x,y,color='teal',alpha=0.2)
plt.plot(x,m*x+b,color='red')
plt.xlabel('total_bill')
plt.ylabel('tip')
plt.title('Linear Regression')
plt.legend()
plt.show()
```
